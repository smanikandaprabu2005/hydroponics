```python
from pathlib import Path
import zipfile
import shutil
import hashlib
import yaml
from collections import Counter


# ============================================================
# CONFIGURATION
# ============================================================

BASE_DIR = Path(r"D:\Hydroponics AI")

# Your existing dataset
CURRENT_DATASET = BASE_DIR / "dataset"

# Roboflow ZIP
ROBOFLOW_ZIP = BASE_DIR / "lettuce yolov11.v14i.yolov11.zip"

# Output dataset
OUTPUT_DATASET = BASE_DIR / "combined_dataset"

# Temporary extraction directory
EXTRACT_DIR = BASE_DIR / "_roboflow_v14_extracted"


# ============================================================
# FINAL 6 CLASSES
# ============================================================

FINAL_CLASSES = {
    0: "Calcium Deficiency",
    1: "Healthy",
    2: "Magnesium Deficiency",
    3: "Nitrogen Deficiency",
    4: "Phosphorus Deficiency",
    5: "Potassium Deficiency",
}


# Roboflow v14 class mapping
# Original ID -> Final ID

ROBOFLOW_CLASS_MAP = {
    0: 1,  # Fully Nutritional -> Healthy
    1: 3,  # Nitrogen -> Nitrogen Deficiency
    2: 4,  # Phosphorus -> Phosphorus Deficiency
    3: 5,  # Potassium -> Potassium Deficiency
}


IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp",
    ".tif",
    ".tiff",
}


# ============================================================
# HELPERS
# ============================================================

def image_hash(path):
    """Calculate SHA256 hash of an image."""
    sha = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(1024 * 1024)

            if not chunk:
                break

            sha.update(chunk)

    return sha.hexdigest()


def find_images(folder):
    """Find all supported images recursively."""
    return [
        p for p in folder.rglob("*")
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ]


def find_label_for_image(image_path):
    """Find corresponding YOLO TXT label."""
    label_path = image_path.with_suffix(".txt")

    if label_path.exists():
        return label_path

    # Search same folder
    alternative = image_path.parent / "labels" / f"{image_path.stem}.txt"

    if alternative.exists():
        return alternative

    return None


def validate_segmentation_label(label_path, num_classes):
    """
    Validate YOLO segmentation label.

    Expected format:

    class x1 y1 x2 y2 x3 y3 ...

    """

    errors = []

    try:
        with open(label_path, "r", encoding="utf-8") as f:
            lines = [
                line.strip()
                for line in f
                if line.strip()
            ]
    except Exception as e:
        return [f"Could not read label: {e}"]

    for line_number, line in enumerate(lines, start=1):

        parts = line.split()

        # class + at least 3 points = 7 values
        if len(parts) < 7:
            errors.append(
                f"Line {line_number}: too few values ({len(parts)})"
            )
            continue

        # Number after class must be even
        if (len(parts) - 1) % 2 != 0:
            errors.append(
                f"Line {line_number}: odd number of polygon coordinates"
            )
            continue

        try:
            class_id = int(parts[0])
        except ValueError:
            errors.append(
                f"Line {line_number}: invalid class ID"
            )
            continue

        if class_id < 0 or class_id >= num_classes:
            errors.append(
                f"Line {line_number}: invalid class ID {class_id}"
            )

        try:
            coordinates = [
                float(x)
                for x in parts[1:]
            ]
        except ValueError:
            errors.append(
                f"Line {line_number}: non-numeric coordinate"
            )
            continue

        for value in coordinates:

            if value < 0 or value > 1:
                errors.append(
                    f"Line {line_number}: coordinate outside [0,1]"
                )
                break

    return errors


def remap_roboflow_label(
    source_label,
    destination_label
):
    """
    Convert Roboflow class IDs to the final 6-class IDs.
    Polygon coordinates are preserved exactly.
    """

    output_lines = []

    with open(source_label, "r", encoding="utf-8") as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            parts = line.split()

            old_class = int(parts[0])

            if old_class not in ROBOFLOW_CLASS_MAP:
                raise ValueError(
                    f"Unexpected Roboflow class ID "
                    f"{old_class} in {source_label}"
                )

            new_class = ROBOFLOW_CLASS_MAP[old_class]

            parts[0] = str(new_class)

            output_lines.append(" ".join(parts))

    with open(destination_label, "w", encoding="utf-8") as f:
        f.write("\n".join(output_lines))

        if output_lines:
            f.write("\n")


def create_output_structure():

    if OUTPUT_DATASET.exists():

        print("\nRemoving existing combined_dataset...")

        shutil.rmtree(OUTPUT_DATASET)

    for split in ["train", "valid", "test"]:

        (OUTPUT_DATASET / split / "images").mkdir(
            parents=True,
            exist_ok=True
        )

        (OUTPUT_DATASET / split / "labels").mkdir(
            parents=True,
            exist_ok=True
        )


def process_current_dataset(
    split,
    hashes,
    statistics
):

    source_image_dir = CURRENT_DATASET / split / "images"
    source_label_dir = CURRENT_DATASET / split / "labels"

    if not source_image_dir.exists():

        print(
            f"WARNING: {source_image_dir} does not exist"
        )

        return

    images = find_images(source_image_dir)

    print(
        f"\nCurrent dataset - {split}: "
        f"{len(images)} images"
    )

    for index, image_path in enumerate(images, start=1):

        label_path = source_label_dir / (
            image_path.stem + ".txt"
        )

        if not label_path.exists():

            print(
                f"WARNING: Missing label: {image_path.name}"
            )

            continue

        # Check segmentation annotation
        errors = validate_segmentation_label(
            label_path,
            len(FINAL_CLASSES)
        )

        if errors:

            print(
                f"\nWARNING: Invalid label: "
                f"{label_path}"
            )

            for error in errors:
                print("   ", error)

            continue

        file_hash = image_hash(image_path)

        # Preserve current dataset if duplicate
        if file_hash in hashes:

            print(
                f"Duplicate image skipped: "
                f"{image_path.name}"
            )

            continue

        hashes[file_hash] = (
            "current",
            split,
            image_path.name
        )

        # Keep original filename, but add current prefix
        new_name = f"current_{index:06d}{image_path.suffix.lower()}"

        destination_image = (
            OUTPUT_DATASET
            / split
            / "images"
            / new_name
        )

        destination_label = (
            OUTPUT_DATASET
            / split
            / "labels"
            / f"current_{index:06d}.txt"
        )

        shutil.copy2(
            image_path,
            destination_image
        )

        shutil.copy2(
            label_path,
            destination_label
        )

        # Count instances
        with open(label_path, "r", encoding="utf-8") as f:

            for line in f:

                line = line.strip()

                if not line:
                    continue

                class_id = int(line.split()[0])

                statistics[split][
                    FINAL_CLASSES[class_id]
                ] += 1


def process_roboflow_dataset(
    split,
    hashes,
    statistics
):

    source_image_dir = (
        EXTRACT_DIR
        / split
        / "images"
    )

    source_label_dir = (
        EXTRACT_DIR
        / split
        / "labels"
    )

    if not source_image_dir.exists():

        print(
            f"\nRoboflow {split} folder not found:"
        )

        print(source_image_dir)

        return

    images = find_images(source_image_dir)

    print(
        f"\nRoboflow v14 - {split}: "
        f"{len(images)} images"
    )

    for index, image_path in enumerate(images, start=1):

        label_path = source_label_dir / (
            image_path.stem + ".txt"
        )

        if not label_path.exists():

            print(
                f"WARNING: Missing Roboflow label: "
                f"{image_path.name}"
            )

            continue

        # Validate original Roboflow segmentation
        errors = validate_segmentation_label(
            label_path,
            4
        )

        if errors:

            print(
                f"\nWARNING: Invalid Roboflow label:"
                f" {label_path}"
            )

            for error in errors:
                print("   ", error)

            continue

        file_hash = image_hash(image_path)

        # Skip exact duplicate
        if file_hash in hashes:

            print(
                f"Duplicate Roboflow image skipped: "
                f"{image_path.name}"
            )

            continue

        hashes[file_hash] = (
            "roboflow",
            split,
            image_path.name
        )

        # Unique filename
        new_name = (
            f"roboflow_{split}_{index:06d}"
            f"{image_path.suffix.lower()}"
        )

        destination_image = (
            OUTPUT_DATASET
            / split
            / "images"
            / new_name
        )

        destination_label = (
            OUTPUT_DATASET
            / split
            / "labels"
            / f"roboflow_{split}_{index:06d}.txt"
        )

        shutil.copy2(
            image_path,
            destination_image
        )

        # Remap class IDs
        remap_roboflow_label(
            label_path,
            destination_label
        )

        # Count final classes
        with open(
            destination_label,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:

                line = line.strip()

                if not line:
                    continue

                class_id = int(line.split()[0])

                statistics[split][
                    FINAL_CLASSES[class_id]
                ] += 1


def write_yaml():

    yaml_data = {
        "path": str(
            OUTPUT_DATASET
        ).replace("\\", "/"),

        "train": "train/images",

        "val": "valid/images",

        "test": "test/images",

        "nc": 6,

        "names": [
            FINAL_CLASSES[i]
            for i in range(6)
        ],
    }

    yaml_path = OUTPUT_DATASET / "data.yaml"

    with open(
        yaml_path,
        "w",
        encoding="utf-8"
    ) as f:

        yaml.safe_dump(
            yaml_data,
            f,
            sort_keys=False,
            allow_unicode=True
        )

    print(
        f"\nCreated: {yaml_path}"
    )


def print_statistics(statistics):

    print("\n")
    print("=" * 70)
    print("FINAL COMBINED DATASET STATISTICS")
    print("=" * 70)

    for split in ["train", "valid", "test"]:

        image_count = len(
            list(
                (
                    OUTPUT_DATASET
                    / split
                    / "images"
                ).glob("*")
            )
        )

        print(f"\n{split.upper()}")
        print("-" * 70)

        print(
            f"Images: {image_count}"
        )

        total_instances = 0

        for class_name in FINAL_CLASSES.values():

            count = statistics[split][class_name]

            total_instances += count

            print(
                f"{class_name:<25}: "
                f"{count}"
            )

        print(
            f"Total instances: {total_instances}"
        )


# ============================================================
# MAIN
# ============================================================
def main():

    print("=" * 70)
    print("YOLO11n-seg DATASET COMBINER")
    print("=" * 70)

    print("\nCurrent dataset:")
    print(CURRENT_DATASET)

    print("\nRoboflow ZIP:")
    print(ROBOFLOW_ZIP)

    print("\nOutput:")
    print(OUTPUT_DATASET)

    # --------------------------------------------------------
    # Check current dataset
    # --------------------------------------------------------

    if not CURRENT_DATASET.exists():
        raise FileNotFoundError(
            f"\nCurrent dataset not found:\n"
            f"{CURRENT_DATASET}"
        )

    # --------------------------------------------------------
    # Check ZIP
    # --------------------------------------------------------

    if not ROBOFLOW_ZIP.exists():
        raise FileNotFoundError(
            f"\nRoboflow ZIP not found:\n"
            f"{ROBOFLOW_ZIP}"
        )

    # --------------------------------------------------------
    # Extract Roboflow dataset
    # --------------------------------------------------------

    if EXTRACT_DIR.exists():
        shutil.rmtree(EXTRACT_DIR)

    EXTRACT_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    print("\nExtracting Roboflow dataset...")

    with zipfile.ZipFile(
        ROBOFLOW_ZIP,
        "r"
    ) as zip_ref:

        zip_ref.extractall(EXTRACT_DIR)

    print("Extraction complete.")

    # --------------------------------------------------------
    # Find actual dataset root
    # --------------------------------------------------------

    roboflow_root = EXTRACT_DIR

    if not (roboflow_root / "train").exists():

        possible_roots = [
            p
            for p in EXTRACT_DIR.iterdir()
            if p.is_dir()
            and (p / "train").exists()
        ]

        if not possible_roots:
            raise FileNotFoundError(
                "Could not find Roboflow train folder "
                "inside the extracted ZIP."
            )

        roboflow_root = possible_roots[0]

    print("\nRoboflow dataset root:")
    print(roboflow_root)

    # --------------------------------------------------------
    # Create output
    # --------------------------------------------------------

    create_output_structure()

    # --------------------------------------------------------
    # Statistics
    # --------------------------------------------------------

    statistics = {
        "train": Counter(),
        "valid": Counter(),
        "test": Counter(),
    }

    # --------------------------------------------------------
    # Global image hashes
    # --------------------------------------------------------

    hashes = {}

    # --------------------------------------------------------
    # Process current dataset
    # --------------------------------------------------------

    print("\n")
    print("=" * 70)
    print("PROCESSING CURRENT DATASET")
    print("=" * 70)

    for split in ["train", "valid", "test"]:

        process_current_dataset(
            split,
            hashes,
            statistics
        )

    # --------------------------------------------------------
    # Process Roboflow dataset
    # --------------------------------------------------------

    print("\n")
    print("=" * 70)
    print("PROCESSING ROBOFLOW v14 DATASET")
    print("=" * 70)

    # Temporarily change EXTRACT_DIR used by the
    # Roboflow processing function.
    original_extract_dir = EXTRACT_DIR

    try:

        # The function expects EXTRACT_DIR to be
        # the actual dataset root.
        globals()["EXTRACT_DIR"] = roboflow_root

        for split in ["train", "valid", "test"]:

            process_roboflow_dataset(
                split,
                hashes,
                statistics
            )

    finally:

        # Restore original value
        globals()["EXTRACT_DIR"] = original_extract_dir

    # --------------------------------------------------------
    # Create data.yaml
    # --------------------------------------------------------

    write_yaml()

    # --------------------------------------------------------
    # Print statistics
    # --------------------------------------------------------

    print_statistics(statistics)

    # --------------------------------------------------------
    # Finish
    # --------------------------------------------------------

    print("\n")
    print("=" * 70)
    print("COMBINATION COMPLETE")
    print("=" * 70)

    print("\nCombined dataset:")
    print(OUTPUT_DATASET)

    print(
        "\nNext step: inspect the combined dataset "
        "before training."
    )
if __name__ == "__main__":
    main()
```

```python
from pathlib import Path
import shutil
import hashlib
import random
from collections import Counter, defaultdict

from PIL import Image
import imagehash
import yaml


# ============================================================
# CONFIGURATION
# ============================================================

BASE_DIR = Path(r"D:\Hydroponics AI")

SOURCE_DATASET = BASE_DIR / "combined_dataset"

OUTPUT_DATASET = BASE_DIR / "combined_dataset_v2"

SEED = 42

TRAIN_RATIO = 0.80
VAL_RATIO = 0.10
TEST_RATIO = 0.10

# Smaller = stricter duplicate detection
# 4 is conservative.
# 6-8 catches more similar images.
PHASH_DISTANCE_THRESHOLD = 5


# ============================================================
# FINAL SIX CLASSES
# ============================================================

CLASSES = {
    0: "Calcium Deficiency",
    1: "Healthy",
    2: "Magnesium Deficiency",
    3: "Nitrogen Deficiency",
    4: "Phosphorus Deficiency",
    5: "Potassium Deficiency",
}


# ============================================================
# IMAGE EXTENSIONS
# ============================================================

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp",
    ".tif",
    ".tiff",
}


# ============================================================
# HELPERS
# ============================================================

def image_hash(path):
    """
    Exact SHA256 image hash.
    Used to detect pixel-identical duplicates.
    """

    sha = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(1024 * 1024)

            if not chunk:
                break

            sha.update(chunk)

    return sha.hexdigest()


def perceptual_hash(path):
    """
    Perceptual hash.

    Similar-looking images produce similar hashes.
    """

    try:

        with Image.open(path) as img:

            img = img.convert("RGB")

            return imagehash.phash(img)

    except Exception as e:

        print(
            f"WARNING: Could not calculate pHash "
            f"for {path.name}: {e}"
        )

        return None


def get_class_ids(label_path):
    """
    Read all class IDs appearing in an image.
    """

    classes = set()

    with open(
        label_path,
        "r",
        encoding="utf-8"
    ) as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            parts = line.split()

            class_id = int(parts[0])

            classes.add(class_id)

    return classes


def get_instance_counts(label_path):
    """
    Count instances of each class in one image.
    """

    counts = Counter()

    with open(
        label_path,
        "r",
        encoding="utf-8"
    ) as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            class_id = int(
                line.split()[0]
            )

            counts[class_id] += 1

    return counts


def find_all_images():

    images = []

    for split in [
        "train",
        "valid",
        "test"
    ]:

        image_dir = (
            SOURCE_DATASET
            / split
            / "images"
        )

        if not image_dir.exists():
            continue

        for path in image_dir.iterdir():

            if (
                path.is_file()
                and path.suffix.lower()
                in IMAGE_EXTENSIONS
            ):

                images.append(path)

    return images


# ============================================================
# CHECK SOURCE DATASET
# ============================================================

if not SOURCE_DATASET.exists():

    raise FileNotFoundError(
        f"Source dataset not found:\n"
        f"{SOURCE_DATASET}"
    )


print("=" * 70)
print("CREATING FINAL TRAIN / VALID / TEST SPLIT")
print("=" * 70)

print("\nSource:")
print(SOURCE_DATASET)

print("\nOutput:")
print(OUTPUT_DATASET)


# ============================================================
# COLLECT ALL IMAGES
# ============================================================

images = find_all_images()

print(
    f"\nTotal images found: {len(images)}"
)


# ============================================================
# BUILD IMAGE RECORDS
# ============================================================

records = []

missing_labels = []

invalid_labels = []

print("\nReading labels...")


for image_path in images:

    label_path = (
        SOURCE_DATASET
        / image_path.parent.parent.name
        / "labels"
        / f"{image_path.stem}.txt"
    )

    # Because parent.parent.name gives train/valid/test
    # in the expected structure.

    if not label_path.exists():

        missing_labels.append(
            image_path
        )

        continue

    try:

        class_ids = get_class_ids(
            label_path
        )

        instance_counts = get_instance_counts(
            label_path
        )

    except Exception as e:

        invalid_labels.append(
            (image_path, str(e))
        )

        continue

    records.append({

        "image": image_path,

        "label": label_path,

        "classes": class_ids,

        "instances": instance_counts,

        "sha256": None,

        "phash": None,

    })


print(
    f"Valid image-label pairs: {len(records)}"
)

print(
    f"Missing labels: {len(missing_labels)}"
)

print(
    f"Invalid labels: {len(invalid_labels)}"
)


# ============================================================
# EXACT DUPLICATE DETECTION
# ============================================================

print("\n")
print("=" * 70)
print("CHECKING EXACT DUPLICATES")
print("=" * 70)

hash_groups = defaultdict(list)

for i, record in enumerate(records):

    record["sha256"] = image_hash(
        record["image"]
    )

    hash_groups[
        record["sha256"]
    ].append(i)


duplicate_groups = [
    group
    for group in hash_groups.values()
    if len(group) > 1
]


duplicate_count = sum(
    len(group) - 1
    for group in duplicate_groups
)


print(
    f"Exact duplicate groups: "
    f"{len(duplicate_groups)}"
)

print(
    f"Duplicate images beyond first copy: "
    f"{duplicate_count}"
)


# ============================================================
# KEEP ONLY ONE COPY OF EXACT DUPLICATES
# ============================================================

unique_records = []

seen_hashes = set()

for record in records:

    h = record["sha256"]

    if h in seen_hashes:

        continue

    seen_hashes.add(h)

    unique_records.append(record)


records = unique_records

print(
    f"Images after exact duplicate removal: "
    f"{len(records)}"
)


# ============================================================
# PERCEPTUAL HASH
# ============================================================

print("\n")
print("=" * 70)
print("CALCULATING PERCEPTUAL HASHES")
print("=" * 70)

for i, record in enumerate(records):

    record["phash"] = perceptual_hash(
        record["image"]
    )

    if (i + 1) % 100 == 0:

        print(
            f"Processed {i + 1}/"
            f"{len(records)}"
        )


# ============================================================
# GROUP NEAR-DUPLICATES
# ============================================================

print("\n")
print("=" * 70)
print("GROUPING NEAR-DUPLICATES")
print("=" * 70)

n = len(records)

parent = list(range(n))


def find(x):

    while parent[x] != x:

        parent[x] = parent[
            parent[x]
        ]

        x = parent[x]

    return x


def union(a, b):

    root_a = find(a)

    root_b = find(b)

    if root_a != root_b:

        parent[root_b] = root_a


near_duplicate_pairs = 0


# Compare pHashes.
# For a few thousand images this is manageable.

for i in range(n):

    hash_i = records[i]["phash"]

    if hash_i is None:
        continue

    for j in range(i + 1, n):

        hash_j = records[j]["phash"]

        if hash_j is None:
            continue

        distance = hash_i - hash_j

        if distance <= PHASH_DISTANCE_THRESHOLD:

            union(i, j)

            near_duplicate_pairs += 1


# ============================================================
# BUILD GROUPS
# ============================================================

groups = defaultdict(list)

for i in range(n):

    root = find(i)

    groups[root].append(i)


groups = list(groups.values())


print(
    f"Total image groups: {len(groups)}"
)

print(
    f"Near-duplicate pairs detected: "
    f"{near_duplicate_pairs}"
)

large_groups = [
    g for g in groups
    if len(g) > 1
]

print(
    f"Groups containing near-duplicates: "
    f"{len(large_groups)}"
)


# ============================================================
# GROUP CLASS INFORMATION
# ============================================================

group_info = []

for group in groups:

    class_counts = Counter()

    for index in group:

        class_counts.update(
            records[index]["instances"]
        )

    group_info.append({

        "indices": group,

        "size": len(group),

        "class_counts": class_counts,

    })


# ============================================================
# RANDOMIZE GROUP ORDER
# ============================================================

random.seed(SEED)

random.shuffle(group_info)


# ============================================================
# TARGET COUNTS
# ============================================================

total_images = len(records)

target_train = int(
    total_images * TRAIN_RATIO
)

target_val = int(
    total_images * VAL_RATIO
)

target_test = (
    total_images
    - target_train
    - target_val
)


print("\nTarget split:")

print(
    f"Train: {target_train}"
)

print(
    f"Valid: {target_val}"
)

print(
    f"Test : {target_test}"
)


# ============================================================
# CLASS-AWARE GROUP ASSIGNMENT
# ============================================================

splits = {

    "train": [],

    "valid": [],

    "test": [],

}


split_sizes = {

    "train": 0,

    "valid": 0,

    "test": 0,

}


# Overall class totals

total_class_counts = Counter()

for record in records:

    total_class_counts.update(
        record["instances"]
    )


# Desired class counts

target_class_counts = {

    "train": Counter(),

    "valid": Counter(),

    "test": Counter(),

}


for class_id in CLASSES:

    total = total_class_counts[class_id]

    target_class_counts["train"][class_id] = (
        total * TRAIN_RATIO
    )

    target_class_counts["valid"][class_id] = (
        total * VAL_RATIO
    )

    target_class_counts["test"][class_id] = (
        total * TEST_RATIO
    )


def split_score(
    split_name,
    group
):

    # Image-size penalty
    size_after = (
        split_sizes[split_name]
        + group["size"]
    )

    if split_name == "train":

        target_size = target_train

    elif split_name == "valid":

        target_size = target_val

    else:

        target_size = target_test

    size_error = abs(
        size_after - target_size
    ) / max(target_size, 1)


    # Class distribution penalty

    class_error = 0.0

    for class_id in CLASSES:

        current = sum(
            records[i]["instances"].get(
                class_id,
                0
            )
            for i in splits[split_name]
        )

        proposed = (
            current
            + group["class_counts"].get(
                class_id,
                0
            )
        )

        target = target_class_counts[
            split_name
        ][class_id]

        if target > 0:

            class_error += abs(
                proposed - target
            ) / target


    # Weight class balance more than image count

    return (
        size_error
        + 0.5 * class_error
    )


# Large groups first

group_info.sort(
    key=lambda x: x["size"],
    reverse=True
)


for group in group_info:

    candidates = []

    for split_name in [
        "train",
        "valid",
        "test"
    ]:

        candidates.append(
            (
                split_score(
                    split_name,
                    group
                ),
                split_name
            )
        )

    candidates.sort()

    selected_split = candidates[0][1]

    splits[selected_split].extend(
        group["indices"]
    )

    split_sizes[
        selected_split
    ] += group["size"]


# ============================================================
# CREATE OUTPUT DIRECTORIES
# ============================================================

if OUTPUT_DATASET.exists():

    print(
        "\nRemoving previous "
        "combined_dataset_v2..."
    )

    shutil.rmtree(
        OUTPUT_DATASET
    )


for split_name in [
    "train",
    "valid",
    "test"
]:

    (
        OUTPUT_DATASET
        / split_name
        / "images"
    ).mkdir(
        parents=True,
        exist_ok=True
    )

    (
        OUTPUT_DATASET
        / split_name
        / "labels"
    ).mkdir(
        parents=True,
        exist_ok=True
    )


# ============================================================
# COPY FILES
# ============================================================

print("\n")
print("=" * 70)
print("COPYING FILES")
print("=" * 70)


split_statistics = {

    "train": Counter(),

    "valid": Counter(),

    "test": Counter(),

}


for split_name, indices in splits.items():

    print(
        f"\nCopying {split_name}: "
        f"{len(indices)} images"
    )

    for counter, index in enumerate(
        indices,
        start=1
    ):

        record = records[index]

        image_path = record["image"]

        label_path = record["label"]

        # Unique names
        new_stem = (
            f"img_{counter:06d}"
        )

        destination_image = (
            OUTPUT_DATASET
            / split_name
            / "images"
            / (
                new_stem
                + image_path.suffix.lower()
            )
        )

        destination_label = (
            OUTPUT_DATASET
            / split_name
            / "labels"
            / (
                new_stem
                + ".txt"
            )
        )

        shutil.copy2(
            image_path,
            destination_image
        )

        shutil.copy2(
            label_path,
            destination_label
        )

        split_statistics[
            split_name
        ].update(
            record["instances"]
        )


# ============================================================
# WRITE DATA.YAML
# ============================================================

yaml_data = {

    "path": str(
        OUTPUT_DATASET
    ).replace("\\", "/"),

    "train": "train/images",

    "val": "valid/images",

    "test": "test/images",

    "nc": 6,

    "names": [
        CLASSES[i]
        for i in range(6)
    ],

}


yaml_path = (
    OUTPUT_DATASET
    / "data.yaml"
)


with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        yaml_data,
        f,
        sort_keys=False,
        allow_unicode=True
    )


# ============================================================
# FINAL REPORT
# ============================================================

print("\n")
print("=" * 70)
print("FINAL DATASET SPLIT")
print("=" * 70)


for split_name in [
    "train",
    "valid",
    "test"
]:

    image_count = len(
        list(
            (
                OUTPUT_DATASET
                / split_name
                / "images"
            ).glob("*")
        )
    )

    print("\n" + split_name.upper())
    print("-" * 70)

    print(
        f"Images: {image_count}"
    )

    total_instances = 0

    for class_id, class_name in CLASSES.items():

        count = split_statistics[
            split_name
        ][class_id]

        total_instances += count

        print(
            f"{class_name:<25}: "
            f"{count}"
        )

    print(
        f"Total instances: "
        f"{total_instances}"
    )


print("\n")
print("=" * 70)
print("CLASS DISTRIBUTION")
print("=" * 70)


for class_id, class_name in CLASSES.items():

    total = sum(
        split_statistics[s][class_id]
        for s in [
            "train",
            "valid",
            "test"
        ]
    )

    print(
        f"\n{class_name}"
    )

    for split_name in [
        "train",
        "valid",
        "test"
    ]:

        count = split_statistics[
            split_name
        ][class_id]

        percentage = (
            100 * count / total
            if total > 0
            else 0
        )

        print(
            f"  {split_name:<6}: "
            f"{count:>6} "
            f"({percentage:5.1f}%)"
        )


print("\n")
print("=" * 70)
print("DONE")
print("=" * 70)

print(
    f"\nFinal dataset:"
)

print(
    OUTPUT_DATASET
)

print(
    f"\nYAML:"
)

print(
    yaml_path
)

print(
    "\nIMPORTANT:"
)

print(
    "Do not train yet. Inspect the final "
    "statistics and sample images first."
)
```

```python
from pathlib import Path
import shutil
import hashlib
import random
from collections import Counter, defaultdict

from PIL import Image
import imagehash
import yaml


# ============================================================
# CONFIGURATION
# ============================================================

BASE_DIR = Path(r"D:\Hydroponics AI")

SOURCE_DATASET = BASE_DIR / "combined_dataset"

OUTPUT_DATASET = BASE_DIR / "combined_dataset_v3"

SEED = 42

TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

PHASH_DISTANCE_THRESHOLD = 5

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp",
    ".tif",
    ".tiff",
}


# ============================================================
# FINAL CLASSES
# ============================================================

CLASSES = {
    0: "Calcium Deficiency",
    1: "Healthy",
    2: "Magnesium Deficiency",
    3: "Nitrogen Deficiency",
    4: "Phosphorus Deficiency",
    5: "Potassium Deficiency",
}


# ============================================================
# HELPER FUNCTIONS
# ============================================================

def exact_hash(path):

    sha = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(1024 * 1024)

            if not chunk:
                break

            sha.update(chunk)

    return sha.hexdigest()


def perceptual_hash(path):

    try:

        with Image.open(path) as img:

            return imagehash.phash(
                img.convert("RGB")
            )

    except Exception as e:

        print(
            f"WARNING: pHash failed for "
            f"{path.name}: {e}"
        )

        return None


def read_label(label_path):

    classes = set()
    counts = Counter()

    with open(
        label_path,
        "r",
        encoding="utf-8"
    ) as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            parts = line.split()

            if len(parts) < 7:
                raise ValueError(
                    f"Invalid segmentation label: "
                    f"{label_path}"
                )

            class_id = int(parts[0])

            if class_id not in CLASSES:
                raise ValueError(
                    f"Invalid class {class_id}: "
                    f"{label_path}"
                )

            # Check polygon coordinate count
            if (len(parts) - 1) % 2 != 0:
                raise ValueError(
                    f"Invalid polygon coordinates: "
                    f"{label_path}"
                )

            # Check coordinates
            for value in parts[1:]:

                value = float(value)

                if not 0 <= value <= 1:

                    raise ValueError(
                        f"Coordinate outside [0,1]: "
                        f"{label_path}"
                    )

            classes.add(class_id)
            counts[class_id] += 1

    return classes, counts


def collect_images():

    records = []

    for split in [
        "train",
        "valid",
        "test"
    ]:

        image_dir = (
            SOURCE_DATASET
            / split
            / "images"
        )

        label_dir = (
            SOURCE_DATASET
            / split
            / "labels"
        )

        if not image_dir.exists():
            continue

        for image_path in image_dir.iterdir():

            if (
                not image_path.is_file()
                or image_path.suffix.lower()
                not in IMAGE_EXTENSIONS
            ):
                continue

            label_path = (
                label_dir
                / f"{image_path.stem}.txt"
            )

            if not label_path.exists():

                print(
                    f"WARNING: Missing label: "
                    f"{image_path.name}"
                )

                continue

            try:

                class_ids, counts = read_label(
                    label_path
                )

            except Exception as e:

                print(
                    f"WARNING: {e}"
                )

                continue

            records.append({

                "image": image_path,

                "label": label_path,

                "classes": class_ids,

                "counts": counts,

                "sha": None,

                "phash": None,

                "group": None,

            })

    return records


# ============================================================
# START
# ============================================================

print("=" * 70)
print("CREATING COMBINED DATASET V3")
print("=" * 70)

print("\nSource:")
print(SOURCE_DATASET)

print("\nOutput:")
print(OUTPUT_DATASET)


if not SOURCE_DATASET.exists():

    raise FileNotFoundError(
        f"Source dataset not found:\n"
        f"{SOURCE_DATASET}"
    )


# ============================================================
# COLLECT RECORDS
# ============================================================

records = collect_images()

print(
    f"\nValid image-label pairs: "
    f"{len(records)}"
)


# ============================================================
# EXACT DUPLICATE GROUPING
# ============================================================

print("\n")
print("=" * 70)
print("EXACT DUPLICATE CHECK")
print("=" * 70)

hash_groups = defaultdict(list)

for i, record in enumerate(records):

    record["sha"] = exact_hash(
        record["image"]
    )

    hash_groups[
        record["sha"]
    ].append(i)


exact_duplicate_groups = [
    group
    for group in hash_groups.values()
    if len(group) > 1
]


print(
    f"Exact duplicate groups: "
    f"{len(exact_duplicate_groups)}"
)


# Keep first copy only

unique_records = []

seen_hashes = set()

for record in records:

    if record["sha"] in seen_hashes:
        continue

    seen_hashes.add(
        record["sha"]
    )

    unique_records.append(record)


records = unique_records

print(
    f"Images after exact duplicate removal: "
    f"{len(records)}"
)


# ============================================================
# PERCEPTUAL HASH
# ============================================================

print("\n")
print("=" * 70)
print("PERCEPTUAL HASH")
print("=" * 70)

for i, record in enumerate(records):

    record["phash"] = perceptual_hash(
        record["image"]
    )

    if (i + 1) % 250 == 0:

        print(
            f"{i + 1}/{len(records)} processed"
        )


# ============================================================
# UNION-FIND FOR NEAR-DUPLICATES
# ============================================================

n = len(records)

parent = list(range(n))


def find(x):

    while parent[x] != x:

        parent[x] = parent[
            parent[x]
        ]

        x = parent[x]

    return x


def union(a, b):

    ra = find(a)
    rb = find(b)

    if ra != rb:

        parent[rb] = ra


near_pairs = 0


for i in range(n):

    hi = records[i]["phash"]

    if hi is None:
        continue

    for j in range(i + 1, n):

        hj = records[j]["phash"]

        if hj is None:
            continue

        distance = hi - hj

        if distance <= PHASH_DISTANCE_THRESHOLD:

            union(i, j)

            near_pairs += 1


# ============================================================
# BUILD GROUPS
# ============================================================

group_dict = defaultdict(list)

for i in range(n):

    root = find(i)

    group_dict[root].append(i)


groups = list(
    group_dict.values()
)


for group_id, group in enumerate(groups):

    for index in group:

        records[index]["group"] = group_id


print(
    f"\nNear-duplicate pairs: "
    f"{near_pairs}"
)

print(
    f"Total independent image groups: "
    f"{len(groups)}"
)


# ============================================================
# GROUP STATISTICS
# ============================================================

group_records = []

for group_id, indices in enumerate(groups):

    class_counts = Counter()

    for index in indices:

        class_counts.update(
            records[index]["counts"]
        )

    group_records.append({

        "id": group_id,

        "indices": indices,

        "size": len(indices),

        "class_counts": class_counts,

    })


# ============================================================
# TOTAL CLASS COUNTS
# ============================================================

total_class_counts = Counter()

for record in records:

    total_class_counts.update(
        record["counts"]
    )


print("\n")
print("=" * 70)
print("TOTAL CLASS INSTANCES")
print("=" * 70)

for class_id, class_name in CLASSES.items():

    print(
        f"{class_name:<25}: "
        f"{total_class_counts[class_id]}"
    )


# ============================================================
# TARGETS
# ============================================================

total_images = len(records)

target_images = {

    "train": total_images * TRAIN_RATIO,

    "valid": total_images * VAL_RATIO,

    "test": total_images * TEST_RATIO,

}


target_classes = {

    "train": Counter(),

    "valid": Counter(),

    "test": Counter(),

}


for class_id in CLASSES:

    total = total_class_counts[class_id]

    target_classes["train"][class_id] = (
        total * TRAIN_RATIO
    )

    target_classes["valid"][class_id] = (
        total * VAL_RATIO
    )

    target_classes["test"][class_id] = (
        total * TEST_RATIO
    )


# ============================================================
# GREEDY GROUP-AWARE STRATIFIED ASSIGNMENT
# ============================================================

random.seed(SEED)

random.shuffle(
    group_records
)


# Process larger groups first

group_records.sort(
    key=lambda g: g["size"],
    reverse=True
)


split_indices = {

    "train": [],

    "valid": [],

    "test": [],

}


split_images = {

    "train": 0,

    "valid": 0,

    "test": 0,

}


split_classes = {

    "train": Counter(),

    "valid": Counter(),

    "test": Counter(),

}


def calculate_score(
    split_name,
    group
):

    # --------------------------------------------
    # Image count error
    # --------------------------------------------

    proposed_images = (
        split_images[split_name]
        + group["size"]
    )

    target_image_count = (
        target_images[split_name]
    )

    image_error = (
        abs(
            proposed_images
            - target_image_count
        )
        / target_image_count
    )


    # --------------------------------------------
    # Class distribution error
    # --------------------------------------------

    class_error = 0

    for class_id in CLASSES:

        current = split_classes[
            split_name
        ][class_id]

        added = group[
            "class_counts"
        ].get(
            class_id,
            0
        )

        proposed = current + added

        target = target_classes[
            split_name
        ][class_id]

        if target > 0:

            class_error += (
                abs(
                    proposed
                    - target
                )
                / target
            )


    # --------------------------------------------
    # Penalize putting too many images
    # --------------------------------------------

    overflow = max(
        0,
        proposed_images
        - target_image_count
    )

    overflow_penalty = (
        10 * overflow
        / max(
            target_image_count,
            1
        )
    )


    return (
        image_error
        + class_error
        + overflow_penalty
    )


# ============================================================
# ASSIGN GROUPS
# ============================================================

for group in group_records:

    candidates = []

    for split_name in [
        "train",
        "valid",
        "test"
    ]:

        score = calculate_score(
            split_name,
            group
        )

        candidates.append(
            (
                score,
                split_name
            )
        )

    candidates.sort()

    selected_split = candidates[0][1]

    split_indices[
        selected_split
    ].extend(
        group["indices"]
    )

    split_images[
        selected_split
    ] += group["size"]

    split_classes[
        selected_split
    ].update(
        group["class_counts"]
    )


# ============================================================
# CREATE OUTPUT
# ============================================================

if OUTPUT_DATASET.exists():

    print(
        "\nRemoving old v3 dataset..."
    )

    shutil.rmtree(
        OUTPUT_DATASET
    )


for split_name in [
    "train",
    "valid",
    "test"
]:

    (
        OUTPUT_DATASET
        / split_name
        / "images"
    ).mkdir(
        parents=True,
        exist_ok=True
    )

    (
        OUTPUT_DATASET
        / split_name
        / "labels"
    ).mkdir(
        parents=True,
        exist_ok=True
    )


# ============================================================
# COPY
# ============================================================

print("\n")
print("=" * 70)
print("COPYING DATA")
print("=" * 70)


final_statistics = {

    "train": Counter(),

    "valid": Counter(),

    "test": Counter(),

}


for split_name in [
    "train",
    "valid",
    "test"
]:

    indices = split_indices[
        split_name
    ]

    print(
        f"\n{split_name}: "
        f"{len(indices)} images"
    )

    for number, index in enumerate(
        indices,
        start=1
    ):

        record = records[index]

        new_stem = (
            f"{split_name}_{number:06d}"
        )

        destination_image = (
            OUTPUT_DATASET
            / split_name
            / "images"
            / (
                new_stem
                + record["image"].suffix.lower()
            )
        )

        destination_label = (
            OUTPUT_DATASET
            / split_name
            / "labels"
            / (
                new_stem
                + ".txt"
            )
        )

        shutil.copy2(
            record["image"],
            destination_image
        )

        shutil.copy2(
            record["label"],
            destination_label
        )

        final_statistics[
            split_name
        ].update(
            record["counts"]
        )


# ============================================================
# DATA.YAML
# ============================================================

yaml_data = {

    "path": str(
        OUTPUT_DATASET
    ).replace("\\", "/"),

    "train": "train/images",

    "val": "valid/images",

    "test": "test/images",

    "nc": 6,

    "names": [
        CLASSES[i]
        for i in range(6)
    ],

}


yaml_path = (
    OUTPUT_DATASET
    / "data.yaml"
)


with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        yaml_data,
        f,
        sort_keys=False,
        allow_unicode=True
    )


# ============================================================
# FINAL REPORT
# ============================================================

print("\n")
print("=" * 70)
print("FINAL DATASET V3")
print("=" * 70)


for split_name in [
    "train",
    "valid",
    "test"
]:

    image_count = len(
        list(
            (
                OUTPUT_DATASET
                / split_name
                / "images"
            ).glob("*")
        )
    )

    total_instances = sum(
        final_statistics[
            split_name
        ].values()
    )

    print(
        f"\n{split_name.upper()}"
    )

    print(
        "-" * 70
    )

    print(
        f"Images: {image_count}"
    )

    print(
        f"Instances: {total_instances}"
    )

    for class_id, class_name in CLASSES.items():

        count = final_statistics[
            split_name
        ][class_id]

        print(
            f"{class_name:<25}: "
            f"{count}"
        )


# ============================================================
# PERCENTAGES
# ============================================================

print("\n")
print("=" * 70)
print("CLASS SPLIT PERCENTAGES")
print("=" * 70)


for class_id, class_name in CLASSES.items():

    total = sum(
        final_statistics[s][class_id]
        for s in [
            "train",
            "valid",
            "test"
        ]
    )

    print(
        f"\n{class_name}"
    )

    for split_name in [
        "train",
        "valid",
        "test"
    ]:

        count = final_statistics[
            split_name
        ][class_id]

        percentage = (
            100 * count / total
            if total > 0
            else 0
        )

        print(
            f"  {split_name:<6}: "
            f"{count:>6} "
            f"({percentage:5.1f}%)"
        )


print("\n")
print("=" * 70)
print("V3 COMPLETE")
print("=" * 70)

print(
    "\nDataset:"
)

print(
    OUTPUT_DATASET
)

print(
    "\nYAML:"
)

print(
    yaml_path
)

print(
    "\nDO NOT TRAIN YET."
)

print(
    "Inspect the statistics and sample images first."
)
```

```python
import os
import shutil
import random
import hashlib
from pathlib import Path
from collections import defaultdict, Counter

import numpy as np
from PIL import Image
import imagehash


# ============================================================
# CONFIGURATION
# ============================================================

SOURCE = Path(r"D:\Hydroponics AI\combined_dataset")
OUTPUT = Path(r"D:\Hydroponics AI\combined_dataset_final")

SEED = 42

# Desired split proportions
TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

# Images with pHash distance <= this value are considered
# near-duplicates and kept in the same split.
PHASH_THRESHOLD = 5

CLASS_NAMES = [
    "Calcium Deficiency",
    "Healthy",
    "Magnesium Deficiency",
    "Nitrogen Deficiency",
    "Phosphorus Deficiency",
    "Potassium Deficiency"
]

NUM_CLASSES = len(CLASS_NAMES)

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}


# ============================================================
# RANDOM SEED
# ============================================================

random.seed(SEED)
np.random.seed(SEED)


# ============================================================
# HELPERS
# ============================================================

def get_image_files(folder):
    return sorted([
        p for p in folder.iterdir()
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ])


def sha256_file(path):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(1024 * 1024)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def calculate_phash(path):
    try:
        img = Image.open(path).convert("RGB")
        return imagehash.phash(img)
    except Exception:
        return None


def read_label(label_path):
    """
    Reads YOLO segmentation labels.

    Each line:
    class_id x1 y1 x2 y2 ... xn yn
    """

    counts = Counter()

    if not label_path.exists():
        return counts

    with open(label_path, "r") as f:
        for line in f:
            parts = line.strip().split()

            if len(parts) < 7:
                continue

            try:
                class_id = int(parts[0])

                if 0 <= class_id < NUM_CLASSES:
                    counts[class_id] += 1

            except ValueError:
                continue

    return counts


# ============================================================
# LOAD IMAGE + LABEL INFORMATION
# ============================================================

print("=" * 70)
print("FINAL GROUP-AWARE DATASET SPLITTER")
print("=" * 70)

print("\nSource:", SOURCE)
print("Output:", OUTPUT)

all_items = []

for split in ["train", "valid", "val", "test"]:

    img_dir = SOURCE / split / "images"
    lbl_dir = SOURCE / split / "labels"

    if not img_dir.exists():
        continue

    for img_path in get_image_files(img_dir):

        label_path = lbl_dir / f"{img_path.stem}.txt"

        class_counts = read_label(label_path)

        all_items.append({
            "image": img_path,
            "label": label_path,
            "class_counts": class_counts,
            "sha256": None,
            "phash": None
        })


print("\nImages found:", len(all_items))


# ============================================================
# EXACT DUPLICATE GROUPING
# ============================================================

print("\nChecking exact duplicates...")

hash_groups = defaultdict(list)

for i, item in enumerate(all_items):

    item["sha256"] = sha256_file(item["image"])

    hash_groups[item["sha256"]].append(i)


exact_duplicate_groups = [
    group for group in hash_groups.values()
    if len(group) > 1
]

print("Exact duplicate groups:", len(exact_duplicate_groups))


# ============================================================
# REMOVE EXACT DUPLICATES
# ============================================================

# Keep only one image from each exact duplicate group.

keep_indices = []

for group in hash_groups.values():
    keep_indices.append(group[0])

keep_indices = sorted(keep_indices)

all_items = [all_items[i] for i in keep_indices]

print("Images after exact duplicate removal:", len(all_items))


# ============================================================
# PHASH CALCULATION
# ============================================================

print("\nCalculating perceptual hashes...")

for i, item in enumerate(all_items):

    item["phash"] = calculate_phash(item["image"])

    if (i + 1) % 250 == 0:
        print(f"  Processed {i + 1}/{len(all_items)}")


# ============================================================
# UNION-FIND FOR NEAR DUPLICATE GROUPS
# ============================================================

class UnionFind:

    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, x):

        while self.parent[x] != x:
            self.parent[x] = self.parent[self.parent[x]]
            x = self.parent[x]

        return x

    def union(self, a, b):

        ra = self.find(a)
        rb = self.find(b)

        if ra == rb:
            return

        if self.rank[ra] < self.rank[rb]:
            self.parent[ra] = rb

        elif self.rank[ra] > self.rank[rb]:
            self.parent[rb] = ra

        else:
            self.parent[rb] = ra
            self.rank[ra] += 1


n = len(all_items)

uf = UnionFind(n)

print("\nFinding near-duplicate groups...")

near_duplicate_pairs = 0

for i in range(n):

    if all_items[i]["phash"] is None:
        continue

    for j in range(i + 1, n):

        if all_items[j]["phash"] is None:
            continue

        distance = all_items[i]["phash"] - all_items[j]["phash"]

        if distance <= PHASH_THRESHOLD:

            uf.union(i, j)
            near_duplicate_pairs += 1


# ============================================================
# BUILD GROUPS
# ============================================================

groups_dict = defaultdict(list)

for i in range(n):
    root = uf.find(i)
    groups_dict[root].append(i)

groups = list(groups_dict.values())

print("Near-duplicate pairs:", near_duplicate_pairs)
print("Independent image groups:", len(groups))


# ============================================================
# GROUP STATISTICS
# ============================================================

group_info = []

for group_id, indices in enumerate(groups):

    class_counts = np.zeros(NUM_CLASSES, dtype=float)

    for idx in indices:

        for cls, count in all_items[idx]["class_counts"].items():
            class_counts[cls] += count

    group_info.append({
        "id": group_id,
        "indices": indices,
        "num_images": len(indices),
        "class_counts": class_counts
    })


# ============================================================
# TOTAL DATASET STATISTICS
# ============================================================

total_images = len(all_items)

total_classes = np.zeros(NUM_CLASSES)

for item in all_items:

    for cls, count in item["class_counts"].items():
        total_classes[cls] += count


print("\n" + "=" * 70)
print("TOTAL DATASET")
print("=" * 70)

print("Images:", total_images)

for i, name in enumerate(CLASS_NAMES):
    print(f"{name:<25}: {int(total_classes[i])}")


# ============================================================
# TARGETS
# ============================================================

ratios = np.array([
    TRAIN_RATIO,
    VAL_RATIO,
    TEST_RATIO
])

target_images = ratios * total_images
target_classes = np.outer(ratios, total_classes)


print("\nTarget image counts:")
print("Train :", round(target_images[0]))
print("Valid :", round(target_images[1]))
print("Test  :", round(target_images[2]))


# ============================================================
# SORT GROUPS
# ============================================================

# Large / class-specific groups are assigned first.
# This gives the optimizer more control over difficult classes.

def group_priority(g):

    counts = g["class_counts"]

    total = counts.sum()

    if total == 0:
        return 0

    # Prioritize groups containing rarer classes.
    rarity_weight = 1.0 / (total_classes + 1)

    score = np.sum(counts * rarity_weight)

    return total * (1 + score)


random.shuffle(group_info)

group_info.sort(
    key=group_priority,
    reverse=True
)


# ============================================================
# SPLIT STATE
# ============================================================

split_names = ["train", "valid", "test"]

split_groups = {
    "train": [],
    "valid": [],
    "test": []
}

split_images = np.zeros(3)
split_classes = np.zeros((3, NUM_CLASSES))


# ============================================================
# COST FUNCTION
# ============================================================

def calculate_cost(split_idx, group):

    new_images = split_images.copy()
    new_classes = split_classes.copy()

    new_images[split_idx] += group["num_images"]
    new_classes[split_idx] += group["class_counts"]

    # --------------------------------------------------------
    # Image distribution error
    # --------------------------------------------------------

    image_error = np.sum(
        ((new_images - target_images) /
         np.maximum(target_images, 1)) ** 2
    )

    # --------------------------------------------------------
    # Class distribution error
    # --------------------------------------------------------

    class_error = 0

    for s in range(3):

        for c in range(NUM_CLASSES):

            target = target_classes[s, c]

            if target > 0:

                class_error += (
                    (new_classes[s, c] - target) / target
                ) ** 2

    # --------------------------------------------------------
    # Penalty if a class is absent from a split
    # --------------------------------------------------------

    missing_penalty = 0

    for s in range(3):

        for c in range(NUM_CLASSES):

            if target_classes[s, c] > 0:

                if new_classes[s, c] == 0:
                    missing_penalty += 5


    # Class balance is more important than exact image ratio.
    return (
        1.0 * image_error +
        4.0 * class_error +
        20.0 * missing_penalty
    )


# ============================================================
# GREEDY GROUP ASSIGNMENT
# ============================================================

print("\nAssigning groups...")

for counter, group in enumerate(group_info):

    costs = []

    for split_idx in range(3):

        cost = calculate_cost(split_idx, group)
        costs.append(cost)

    best_split = int(np.argmin(costs))

    split_name = split_names[best_split]

    split_groups[split_name].append(group)

    split_images[best_split] += group["num_images"]
    split_classes[best_split] += group["class_counts"]

    if (counter + 1) % 250 == 0:
        print(f"  Assigned {counter + 1}/{len(group_info)} groups")


# ============================================================
# LOCAL IMPROVEMENT
# ============================================================

print("\nRunning local optimization...")


def current_total_cost():

    image_error = np.sum(
        ((split_images - target_images) /
         np.maximum(target_images, 1)) ** 2
    )

    class_error = 0

    for s in range(3):

        for c in range(NUM_CLASSES):

            target = target_classes[s, c]

            if target > 0:

                class_error += (
                    (split_classes[s, c] - target) / target
                ) ** 2

    missing_penalty = 0

    for s in range(3):

        for c in range(NUM_CLASSES):

            if target_classes[s, c] > 0 and split_classes[s, c] == 0:
                missing_penalty += 5

    return image_error + 4 * class_error + 20 * missing_penalty


# A limited number of swaps/moves prevents excessive runtime.

for iteration in range(10):

    improved = False

    current_cost = current_total_cost()

    # Randomize group order for local search.
    shuffled_groups = group_info.copy()
    random.shuffle(shuffled_groups)

    for group in shuffled_groups[:1000]:

        current_split = None

        for s, name in enumerate(split_names):

            if group in split_groups[name]:

                current_split = s
                break

        if current_split is None:
            continue

        for new_split in range(3):

            if new_split == current_split:
                continue

            # Temporarily move group
            split_groups[split_names[current_split]].remove(group)
            split_groups[split_names[new_split]].append(group)

            split_images[current_split] -= group["num_images"]
            split_images[new_split] += group["num_images"]

            split_classes[current_split] -= group["class_counts"]
            split_classes[new_split] += group["class_counts"]

            new_cost = current_total_cost()

            if new_cost < current_cost:

                current_cost = new_cost
                improved = True
                break

            # Undo move
            split_groups[split_names[new_split]].remove(group)
            split_groups[split_names[current_split]].append(group)

            split_images[current_split] += group["num_images"]
            split_images[new_split] -= group["num_images"]

            split_classes[current_split] += group["class_counts"]
            split_classes[new_split] -= group["class_counts"]

    print(
        f"Optimization iteration {iteration + 1}: "
        f"cost={current_cost:.4f}"
    )

    if not improved:
        break


# ============================================================
# CREATE OUTPUT DIRECTORY
# ============================================================

if OUTPUT.exists():

    print("\nRemoving existing output directory...")

    shutil.rmtree(OUTPUT)


for split in ["train", "valid", "test"]:

    (OUTPUT / split / "images").mkdir(parents=True)
    (OUTPUT / split / "labels").mkdir(parents=True)


# ============================================================
# COPY FILES
# ============================================================

print("\nCopying files...")

for split_name in split_names:

    output_split = OUTPUT / split_name

    for group in split_groups[split_name]:

        for idx in group["indices"]:

            item = all_items[idx]

            shutil.copy2(
                item["image"],
                output_split / "images" / item["image"].name
            )

            if item["label"].exists():

                shutil.copy2(
                    item["label"],
                    output_split / "labels" / item["label"].name
                )


# ============================================================
# CREATE DATA.YAML
# ============================================================

yaml_text = f"""path: {OUTPUT.as_posix()}
train: train/images
val: valid/images
test: test/images

nc: {NUM_CLASSES}

names:
"""

for i, name in enumerate(CLASS_NAMES):
    yaml_text += f"  {i}: {name}\n"


with open(OUTPUT / "data.yaml", "w") as f:
    f.write(yaml_text)


# ============================================================
# FINAL REPORT
# ============================================================

print("\n" + "=" * 70)
print("FINAL SPLIT RESULT")
print("=" * 70)

for s, split_name in enumerate(split_names):

    print(f"\n{split_name.upper()}")

    num_images = int(split_images[s])

    print("Images:", num_images)

    print(
        "Percentage:",
        f"{100 * num_images / total_images:.2f}%"
    )

    print("\nInstances:")

    for c, name in enumerate(CLASS_NAMES):

        count = int(split_classes[s, c])

        percentage = (
            100 * count / total_classes[c]
            if total_classes[c] > 0
            else 0
        )

        print(
            f"  {name:<25} "
            f"{count:>6} "
            f"({percentage:>5.1f}% of class)"
        )


# ============================================================
# CLASS BALANCE TABLE
# ============================================================

print("\n" + "=" * 70)
print("CLASS DISTRIBUTION ACROSS SPLITS")
print("=" * 70)

print(
    f"{'Class':<25}"
    f"{'Train %':>10}"
    f"{'Val %':>10}"
    f"{'Test %':>10}"
)

for c, name in enumerate(CLASS_NAMES):

    percentages = []

    for s in range(3):

        pct = (
            100 * split_classes[s, c] /
            total_classes[c]
        )

        percentages.append(pct)

    print(
        f"{name:<25}"
        f"{percentages[0]:>9.1f}%"
        f"{percentages[1]:>9.1f}%"
        f"{percentages[2]:>9.1f}%"
    )


# ============================================================
# CHECK ALL CLASSES
# ============================================================

print("\n" + "=" * 70)
print("VALIDATION CHECKS")
print("=" * 70)

all_good = True

# Every split must contain every class.

for s, split_name in enumerate(split_names):

    missing = []

    for c, name in enumerate(CLASS_NAMES):

        if split_classes[s, c] == 0:
            missing.append(name)

    if missing:

        all_good = False

        print(
            f"WARNING: {split_name} is missing:",
            missing
        )

    else:

        print(
            f"OK: {split_name} contains all 6 classes"
        )


# Image ratio check

print("\nImage split:")

for s, name in enumerate(split_names):

    actual = 100 * split_images[s] / total_images
    target = 100 * ratios[s]

    print(
        f"{name:<8} "
        f"actual={actual:.2f}% "
        f"target={target:.2f}%"
    )


print("\nOutput:", OUTPUT)

print("\n" + "=" * 70)

if all_good:
    print("FINAL SPLIT CREATED SUCCESSFULLY")
else:
    print("SPLIT CREATED WITH WARNINGS — REVIEW ABOVE")

print("=" * 70)
```
